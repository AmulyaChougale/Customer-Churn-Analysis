```python
sns.countplot(x='Exited', data=ds)
```

```python
churn_count=df['Exited'].value_counts()
churnC=df['Exited'].value_counts(normalize=True)*100
```

```python
labels = ['Stayed', 'Churned']

plt.figure(figsize=(6,6))
plt.pie(churn_count, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title('Customer Churn Distribution')
plt.show()
```

```python
#Churn by Geography
plt.figure(figsize=(10,6))
sns.barplot(x='Geography',y='Exited',data=ds)
plt.title("Churn Rate by Geography")
plt.ylabel("Churn Rate")
plt.xlabel("Country")
plt.show()
```

```python
#Churn Rate by Gender
plt.figure(figsize=(10,6))
sns.barplot(x='Gender',y='Exited',data=ds)
plt.title("Churn Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Churn Rate")
plt.show()
```

```python
#Churn Rate by Age
plt.figure(figsize=(12,8))
sns.boxplot(x='Exited',y='Age',data=ds)
plt.title("Churn Rate by Age")
plt.xlabel("Churn")
plt.ylabel("Age")
plt.show()
```

```python
#Churn Rate by Products
plt.figure(figsize=(8,6))
sns.barplot(x='NumOfProducts',y='Exited',data=ds)
plt.title("Churn Rate by Products")
plt.xlabel("No. of Products")
plt.ylabel("Churn Rate")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.barplot(x='IsActiveMember', y='Exited', data=ds)
plt.title('Churn Rate by Active Membership')
plt.xlabel('Active Member (0 = No, 1 = Yes)')
plt.ylabel('Churn Rate')
plt.show()
```
